```python
import mysql.connector
import os
from dotenv import load_dotenv

load_dotenv()

connection = mysql.connector.connect(
    host=os.getenv("DB_HOST"),
    user=os.getenv("DB_USER"),
    password=os.getenv("DB_PASSWORD"),
    database=os.getenv("DB_NAME")
)

print("Connected to Little Lemon database successfully!")
```

```python
import sys

print(sys.executable)
print(sys.version)
```

```python
cursor = connection.cursor()

print("Cursor created successfully!")
```

```python
cursor.execute("SELECT DATABASE();")

result = cursor.fetchone()

print("Connected to database:", result[0])
```

```python
cursor.close()

cursor = connection.cursor()

cursor.execute("""
      SELECT BookingID, BookingDate, TableNumber, CustomerID
      FROM Bookings
      ORDER by BookingID
""")

bookings = cursor.fetchall() # we use fetchall() to get all the rows returned by the query

for booking in bookings:
    print(booking)
```

```python
show_tables_query = "SHOW TABLES"

cursor.execute(show_tables_query)

results = cursor.fetchall()

print(results)
```

```python
join_query = """
SELECT
    customers.FullName,
    customers.PhoneNumber,
    customers.Email,
    orders.TotalCost
FROM customers
JOIN orders
    ON customers.CustomerID = orders.CustomerID
WHERE orders.TotalCost > 60;
"""

cursor.execute(join_query)

results = cursor.fetchall()

for row in results:
    print(row)
```
